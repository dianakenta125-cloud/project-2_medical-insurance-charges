```python
#import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
#encode categorical variables
df["smoker"]=df["smoker"].map({"yes":1,"no":0})
```

```python
df.head()
```

```python
#select variables
x=df[["age","bmi","children","smoker"]]
y=df["charges"]
```

```python
#justification
age,bmi,children,smoker were selected as independent variables because they influence individual medical insurance
insurance selected as dependent variable because it predicts the cost of medical insurance
```

```python
x.head()
```

```python
y.head()
```

```python
#split data
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

model =LinearRegression()
model.fit(x_train,y_train)

```python
#train Multiple linear Regression model
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
#get intercept
print("intercept:",model.intercept_)
```

```python
#get regression coefficients
coefficients = pd.DataFrame({
    "variable":x.columns,
    "coefficient":model.coef_
})
print(coefficients)
```

```python
#prediction
y_predict=model.predict(x_test)
print(y_predict)
```

```python
#evaluate model
mse=mean_squared_error(y_test,y_predict)
print(mse)
```

```python
for variable,coefficient in zip(x.columns,model.coef_):
    print(variable,":",coefficient)
```

```python
#explanation
age_holding bmi,number of children,and smoking constant,an increase in age is associated with increase in predicted insurance charges 
BMI_holding other variables constant,increase in bmi is associated with change  in predicted insurance charges equal to bmi coefficient
children_holding other variables constant,an additional child is associated with a change in predicted insurance charges equal to the children coefficients
smoker_smoking status from non_smoker to smoker changes predicted insurance charges by the smoker coefficient holding other variables constant.
```
